# 1. 封装（Encapsulation）

	•	定义：将数据和方法包装在一个类中，并隐藏内部实现细节

	•	目的：防止外部直接访问和修改属性，增强安全性与代码可维护性

	•	Python 写法：使用 _ 或 __ 定义私有属性

## 非私有性写法

````python
class LED:
  def __init__(self, name):
    self.name = name
    self.status = False

  def toggle(self):
    before = self.status
    self.status = not self.status  # 通过方法修改
    print(f'{self.name} is toggled from {before} to {self.status}')
````

````python
led_0 = LED("红灯")
led_0.status
````

````python
led_0.toggle()
````

````python
led_0.status
````

## 封装写法

````python
class LED:
  def __init__(self, name):
    self.name = name
    self.__status = False #  私有属性

  def toggle(self):
    before = self.__status
    self.__status = not self.__status  # 只能通过方法修改
    print(f'{self.name} is toggled from {before} to {self.__status}')
````

````python
class LED:
  def __init__(self, name):
    self.name = name
    self._status = False #  保护属性

  def toggle(self):
    before = self._status
    self._status = not self._status  # 只能通过方法修改
    print(f'{self.name} is toggled from {before} to {self._status}')
````

````python
led = LED("红灯")
led.toggle()
````

````python
class Lamp:
    def __init__(self, color):
        self.color = color
        self.__is_on = False  # 私有属性

    def toggle(self):
        self.__is_on = not self.__is_on

    def status(self):
        self.__is_on = not self.__is_on
        return f"{self.color} 灯 状态：{'开' if self.__is_on else '关'}"
````

# 2. 组合 & 聚合（Composition & Aggregation）

	•	组合：类中直接创建和管理子对象

	•	聚合：外部创建对象传入类中使用

````python
class LampGroup:
    def __init__(self):
        self.red = Lamp("红")
        self.green = Lamp("绿")
        self.blue = Lamp("蓝")

    def toggle(self, color):
        if hasattr(self, color):
            getattr(self, color).toggle()
            return f"{color} 灯 状态变了"

    def show_status(self):
        for lamp in [self.red, self.green, self.blue]:
            print(lamp.status())
````

# 3. 总结

````python
# Lamp 类 (封装)
class Lamp:
    def __init__(self, color: str):
        self.color = color
        self.__is_on = False

    def toggle(self):
        self.__is_on = not self.__is_on

    def status(self):
        return f"{self.color} lamp is {'ON' if self.__is_on else 'OFF'}"
````

````python
# LampGroup (组合)
class LampGroup:
    def __init__(self):
        self.red   = Lamp("Red")
        self.green = Lamp("Green")
        self.blue  = Lamp("Blue")

    def toggle(self, color: str):
        lamp = getattr(self, color, None)
        if lamp:
            lamp.toggle()

    def show(self):
        for lamp in (self.red, self.green, self.blue):
            print(lamp.status())
````

````python
# LampGroup (聚合)
# 先定义 Lamp 类（简单封装）
class Lamp:
    def __init__(self, name):
        self.name = name
        self._on = False

    def toggle(self):
        self._on = not self._on

    def status(self):
        return f"{self.name} 状态：{'开' if self._on else '关'}"

# 聚合关系：LampGroup 不拥有灯，而是“接收”它们
class LampGroup:
    def __init__(self, red_lamp, green_lamp, blue_lamp):
        self.lamps = {
            "red": red_lamp,
            "green": green_lamp,
            "blue": blue_lamp
        }

    def toggle(self, color: str):
        lamp = self.lamps.get(color)
        if lamp:
            lamp.toggle()
        else:
            print(" 没有这个颜色的灯！")

    def show(self):
        for lamp in self.lamps.values():
            print(lamp.status())

# 外部先创建灯实例
red   = Lamp("红灯")
green = Lamp("绿灯")
blue  = Lamp("蓝灯")

# 灯传入 LampGroup，形成聚合关系
group = LampGroup(red, green, blue)

# 控制逻辑
group.toggle("red")
group.show()
````

    group = LampGroup()
