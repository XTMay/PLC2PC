## 📌 理论讲解


1. 继承（Inheritance）


```
	•	定义：子类继承父类的属性和方法
	•	优点：代码复用、逻辑抽象、减少重复代码
```



````python
class Sensor:
    def __init__(self, name):
        self.name = name

    def read(self):
        return "ALL Sensor data"
````

````python
class TempSensor(Sensor):
    def read(self):
        return "23.5 °C"
````

````python
class PressureSensor(Sensor):
    def read(self):
        return "101.3 kPa"
````

````python
t = TempSensor("温度传感器")
p = PressureSensor("气压传感器")
print(t.name, "读取:", t.read())
print(p.name, "读取:", p.read())
````

2. 方法重写（Override）


```

	•	子类定义同名方法时会覆盖父类的方法
	•	可在子类中使用 super() 调用父类方法
```



````python
class LoggerSensor(Sensor):
    def read(self): # self: 自己的 ， super： 父亲的
        value = super().read() + " LoggerSensor" # "ALL Sensor data LoggerSensor"
        print(f"[{self.name}] 读取中: {value}")

        return value
````

````python
l = LoggerSensor("万能传感器")
````

````python
l.name
````

````python
l.read()
````

3. 多态（Polymorphism）

```
	•	同一种接口，不同子类表现出不同的行为
	•	通过父类引用调用子类方法，实现统一操作接口
```



````python
def display_readings(sensors):
    for sensor in sensors:
        print(f"{sensor.name}: {sensor.read()}")

sensors = [TempSensor("温度1"), PressureSensor("气压1")]
display_readings(sensors)
````

4. 抽象基类（Abstract Base Class）



```
	•	使用 Python abc 模块定义规范化接口
	•	子类必须实现抽象方法，否则无法实例化
```



````python
from abc import ABC, abstractmethod

class SensorBase(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def read(self):
        pass

class TempSensor(SensorBase):
  def call(self):
    print("call")

  def read(self):
      return "23.5 °C"
````

````python
temp = TempSensor("Temp")
````

````python
temp.name
````

````python
temp.call()
````

````python
temp.read()
````

## 🧪 实作练习

### Part 1：继承与重写

````python
class Sensor:
    def __init__(self, name):
        self.name = name
    def read(self):
        return "Generic Sensor"

class TempSensor(Sensor):
    def read(self):
        return "Temperature: 22.8 °C"

class PressureSensor(Sensor):
    def read(self):
        return "Pressure: 101.2 kPa"
````

### Part 2：统一接口，多态调用

````python
def show_all(sensors):
    for s in sensors:
        print(f"{s.name} -> {s.read()}")

sensor_list = [TempSensor("T1"), PressureSensor("P1")]
show_all(sensor_list)
````
